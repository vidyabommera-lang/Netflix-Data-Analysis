# Netflix Data Analysis

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('netflix_titles.csv')
print(df.head())
print(df.info())
print(df.describe(include='all'))
print(df.isnull().sum())
df=df.drop_duplicates()
df['type'].value_counts().plot(kind='bar')
plt.show()
```
